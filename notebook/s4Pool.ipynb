```python
import torch
import torch.nn as nn
from torch.linalg import eigh, inv, matrix_power
import torch.nn.functional as F
import math
torch.set_printoptions(precision=3, sci_mode=False)

def scan_SSM(Ab, Bb, Cb, u, x0):
    # u.Size(..., seq, dim)
    x_k, y = x0, []
    for u_k in u.unbind(dim=-2):
        x_k = x_k @ Ab.T + u_k.unsqueeze(-1) @ Bb.T
        y.append((x_k @ Cb.T).squeeze())
    return torch.stack(y), x_k

N, dim = 8, 4
Ab, Bb, Cb = torch.randn(N, N), torch.randn(N,1), torch.randn(1,N)
u = torch.randn(2, 3, dim)
x0 = torch.zeros(dim, N)
scan_SSM(Ab, Bb, Cb, u, x0)
```
